```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
heart_failure = pd.read_csv('heart_failure_clinical_records_dataset.csv')
heart_failure.head()
```

```python
heart_failure["ejection_fraction"].plot.kde()
```

```python
for i in range(15):
    sample = heart_failure.sample(frac=1.0, replace=True)
    sample["ejection_fraction"].plot.kde()

plt.xlabel("Ejection Fraction")
plt.ylabel("Density")
plt.show()
```

The original KDE appears relatively reliable between approximately 20 and 30 and between 40 and 50, where the resampled KDE curves are relatively similar. There is noticeable variation around 35–40, so those values may not be as reliable.
